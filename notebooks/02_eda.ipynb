```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/processed/final_cleaned_dataset.csv")

print(df.shape)
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Daily sales
daily_sales = df.groupby("date")["units_sold"].sum()

plt.figure(figsize=(15,5))
plt.plot(daily_sales)

plt.title("Daily Units Sold")
plt.xlabel("Date")
plt.ylabel("Units Sold")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
category_revenue = (
    df.groupby("category")["revenue"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))
category_revenue.plot(kind="bar")

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    df.groupby(df["date"].dt.month)["units_sold"]
      .sum()
)

plt.figure(figsize=(8,5))
monthly_sales.plot(kind="bar")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.tight_layout()
plt.show()
```

```python
promo_sales = df.groupby("promo_flag")["units_sold"].mean()

plt.figure(figsize=(6,4))
promo_sales.plot(kind="bar")

plt.title("Average Units Sold by Promotion")
plt.xlabel("Promotion Flag")
plt.ylabel("Average Units Sold")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

df["inventory_status"].value_counts().plot(kind="bar")

plt.title("Inventory Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

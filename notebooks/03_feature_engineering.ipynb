```python
import pandas as pd

columns = [
    "date",
    "sku_id",
    "units_sold",
    "promo_flag",
    "is_holiday",
    "month",
    "day_of_week",
    "inventory_status",
    "opening_stock",
    "closing_stock"
]

df = pd.read_csv(
    "../data/processed/final_cleaned_dataset.csv",
    usecols=columns
)

print(df.shape)
```

```python
df.info(memory_usage="deep")
```

```python
categorical_cols = [
    "sku_id",
    "promo_flag",
    "is_holiday",
    "month",
    "day_of_week",
    "inventory_status"
]

for col in categorical_cols:
    df[col] = df[col].astype("category")

print(df.info(memory_usage="deep"))
```

```python
df.dtypes
```

```python
df["units_sold"] = df["units_sold"].astype("int32")
df["opening_stock"] = df["opening_stock"].astype("int32")
df["closing_stock"] = df["closing_stock"].astype("int32")
```

```python
df.to_csv("../data/processed/final_features_dataset.csv", index=False)

print("✅ Feature dataset saved successfully!")
```

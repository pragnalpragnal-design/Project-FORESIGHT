```python
sales["sku_id"].nunique()
```

```python
sales_sku = pd.merge(
    sales,
    sku,
    on="sku_id",
    how="left"
)

print(sales_sku.shape)
sales_sku.head()
```

```python
sales_sku_calendar = pd.merge(
    sales_sku,
    calendar,
    on="date",
    how="left"
)

print(sales_sku_calendar.shape)
sales_sku_calendar.head()
```

```python
final_df = pd.merge(
    sales_sku_calendar,
    inventory,
    on=["date", "sku_id"],
    how="left",
    suffixes=("", "_inventory")
)

print(final_df.shape)
```
